# Running small GCMC using GraphCMC

Welcome to the tutorial! In this section, you'll learn how to run a small GCMC simulation using the `graphcmc` package. Currently, this package supports GCMC simulations for CO<sub>2</sub> only. As an example, we'll use Mg-MOF-74 (CSD code: `OPAGIX`) to demonstrate a simple CO<sub>2</sub> adsorption calculation.

### Supported Moves
At this stage, `graphcmc` handles the following GCMC moves with equal probabilities:

- Insertion (25%)
- Deletion (25%)
- Translation (25%)
- Rotation (25%)

Additional flexibility for GCMC move types is planned for the future updates.

## Getting Started

First, import the `graphcmc` package. The main GCMC simulation class can be accessed via:

```python
from graphcmc.simulation import GCMC
```

Now, we'll initialize a `GCMC` object with desired simulation settings. Key arguments include:

- `mofname`: name of the MOF (expects `{mofname}.cif` in the current directory)
- `adsname`: name of the adsorbate (CO<sub>2</sub> only, automatically loads `co2.xyz`)
- `T`: external temperature [K]
- `P`: external pressure [Pa]
- `ff`: force field (`UFF` or any MLFF listed in `README.md`)
- `vdw_cutoff`: van der Waals cutoff (default: 14 Å)
- `device`: `cpu` or `cuda` (default: `cpu`)
- `print_every`: frequency to print simulation stats (per cycle)
- `result_dir`: directory to save outputs (default: `results`)

The `GCMC` object automatically expands the unit cell to avoid self-interactions. It ensures that each lattice vector is at least twice the `vdw_cutoff`.

## Example Setup

Let's simulate CO<sub>2</sub> adsorption in OPAGIX at 298 K and 1000 Pa using UFF. The steps for using MLFFs are identical. Just make sure to set `device=cuda` if running on a GPU.

```python
gcmc = GCMC('OPAGIX', 'co2', T = 298.0, P = 1000.0, ff = 'UFF', print_every = 5)
```

## Initialization Cycles

Begin by running a few initialization cycles. For this tutorial, we'll start with just 10:

```python
gcmc.mc(10, initialize = True)
```

## Production cycles

Once the initialization cycles are finished, a checkpoint file is saved in `results/OPAGIX_co2_298.0K_0001000Pa_UFF/`.

From here, you can run production cycles:

```python
gcmc.mc(10)
```

You can also extend an existing simulation. For example, to add 10 more cycles (for a total of 20), simply call `gcmc.mc()` again with `continue_sim=True`.

```python
gcmc.mc(20, continue_sim = True)
```

In this case, only the additional cycles will be printed in the output.

## Restarting a Simulation

You can restart simulations at various stages:

- Redo only production cycles from the latest initialization checkpoint:

```python
gcmc.mc(20, continue_sim = False)
```

- Redo initialization from scratch:

```python
gcmc.mc(10, initialize = True, continue_sim = False)
```

## Changing Parameters Mid-Simulation

To reset temperature or pressure, you can directly modify the `GCMC` object. The output directory and fugacity will automatically update. Results will be saved in the new output directory.

```python
## former fugacity
print(f'Fugacity (T = 298 K, P = 1000 Pa): {gcmc.fugacity}')

# updated fugacity
gcmc.P = 100
print(f'Fugacity (T = 298 K, P = 100 Pa): {gcmc.fugacity}')
```

```python
gcmc.mc(10, initialize = True)
gcmc.mc(10)
```
